# Removing the timeout generic error

```python
remove_timeout_error_patten = r"Traceback(.*)(?=Traceback)"
```

```python
removed_timeout_error = re.search(remove_timeout_error_patten, error_message, re.DOTALL).group().strip()
print(remove_timeout_error)
```
