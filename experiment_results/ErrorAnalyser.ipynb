```python
import pandas as pd
import os
import ast
```

```python
files = os.listdir(".")
```

```python
files
```

```python
all_dfs = []
used_files = []

for file in files:
    if not file.endswith("_results.csv"):
        continue
        
    df = pd.read_csv(file)
    if df.shape[1] != 4:
        print(f"Skipping {file} — has {df.shape[1]} columns instead of 4")
        continue
    
    name_part = file.replace("_results.csv", "")
    model,benchmark = name_part.split("_")
    df["model"] = model 
    df["benchmark"] = benchmark
    
    all_dfs.append(df)
    used_files.append(file)
```
